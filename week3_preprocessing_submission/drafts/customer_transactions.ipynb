### import libraries and dataset

```python
import pandas as pd
import numpy as np

# load dataset

transactions = pd.read_csv('../../data/raw/customer_transactions_contaminated.csv')
```

### inspect

```python
# inspect dataset
print("First 20 rows: ")
display(transactions.head(20))

print("\nDataset Information: ")
display(transactions.info())

print("\nDataset Dimensions: ")
display(transactions.shape)

print("\nDataset Columns: ")
display(transactions.columns)

print("\nDataset Numerical Statistics: ")
display(transactions.describe())
```

### exact duplicate records

```python
# inspect duplicates
duplicate_transactions = transactions.duplicated().sum()

print("Exact duplicate rows:", duplicate_transactions)

display(transactions[transactions.duplicated(keep=False)])
```

```python
# drop exact duplicate records
transactions_cleaned = transactions.drop_duplicates().copy()

# verify and check shape 
print("Duplicates remaining:", transactions_cleaned.duplicated().sum())
print("Dataset shape after removal of duplicates:", transactions_cleaned.shape)
```

```python
# inspect whether any non-identical records still share a transaction id
print("Unique TransactionIDs:", transactions_cleaned["TransactionID"].nunique())
print("Duplicate TransactionIDs:", transactions_cleaned["TransactionID"].duplicated().sum())
```

```python
# inspect whether those 15 id's are duplicate copies of the same transaction with some corrupted/different field, or genuinely different transactions that were accidentally given the same id
duplicate_transaction_ids = transactions_cleaned[transactions_cleaned["TransactionID"].duplicated(keep=False)].sort_values("TransactionID")
display(duplicate_transaction_ids)
```

```python
# check how many rows each duplicated id has
duplicate_transaction_ids["TransactionID"].value_counts()
```

after removing exact duplicate rows, 15 transaction ids were still associated with two different records. these records were inspected and found to contain differences in fields such as transaction date and amount. since the records are not exact duplicates, they were retained rather than arbitrarily deleting one record from each pair. the inconsistent transaction id values will be documented as data-quality issues.

### missing, invalid, and inconsistent values

#### amount

```python
# inspect amount column distribution
print("Amount Values:")
display(transactions_cleaned["Amount"].value_counts(dropna=False).sort_index())
print("\nAmount Data Type:", transactions_cleaned["Amount"].dtype)
```

```python
# convert amount values to numeric to properly analyze
transactions_cleaned["Amount"] = pd.to_numeric(transactions_cleaned["Amount"], errors="coerce")

# verify
print("Amount Column Data Type:", transactions_cleaned["Amount"].dtype)
print("\nMissing Amount Values:", transactions_cleaned["Amount"].isna().sum())
```
