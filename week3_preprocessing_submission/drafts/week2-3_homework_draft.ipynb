```python
import pandas as pd

# load datasets

customers = pd.read_csv('../../data/raw/customer_demographics_contaminated.csv')
transactions = pd.read_csv('../../data/raw/customer_transactions_contaminated.csv')
interactions = pd.read_csv('../../data/raw/social_media_interactions_contaminated.csv')
```

### inspect and clean customer demographics dataset

```python
# inspect dataset
print("First 20 rows: ")
display(customers.head(20))

print("\nDataset Information: ")
display(customers.info())

print("\nDataset Dimensions: ")
display(customers.shape)

print("\nDataset Columns: ")
display(customers.columns)

print("\nDataset Numerical Statistics: ")
display(customers.describe())
```

##### handling exact duplicate values

```python
# check duplicates
duplicate_customers = customers.duplicated().sum()

print("Number of Exact Duplicate Rows: ")
print(duplicate_customers)

# inspect duplicates
display(customers[customers.duplicated(keep=False)])
```

```python
# remove duplicates
customer_cleaned = customers.drop_duplicates().copy()

# verify
print("Duplicates Remaining:")
print(customer_cleaned.duplicated().sum())
```

#### handling missing and invalid values

##### age column

```python
# check missing values
missing_values = customer_cleaned.isnull().sum()
print("Missing Values:")
print(missing_values)
```

```python
# inspect data types and check numerical statistics
print("Data Types:")
print(customer_cleaned.dtypes)

print("\nNumerical Statistics:")
display(customer_cleaned.describe())
```

```python
# check categorical inconsistencies
for column in customer_cleaned.select_dtypes(include="object").columns:
    print(f"\n{column}")
    print(customer_cleaned[column].value_counts(dropna=False))
```

```python
# convert age from string to integer
customer_cleaned["Age"] = pd.to_numeric(customer_cleaned["Age"], errors="coerce")

# check numerical statistics
print("Numerical Statistics of Age Column:")
display(customer_cleaned["Age"].describe())

# check again number of missing values
print("Number of Missing Values:")
missing_age = customer_cleaned["Age"].isnull().sum().copy()
print(missing_age)
```

invalid age values are dealt with prior to filling or dropping missing values

```python
# count how many invalid age values are in the dataset
invalid_age = customer_cleaned[(customer_cleaned["Age"] < 0) | (customer_cleaned["Age"] > 100)]
print("Invalid Ages:", len(invalid_age))
display(invalid_age)
```

```python
# convert invalid values to missing values
customer_cleaned.loc[(customer_cleaned["Age"] < 0) | (customer_cleaned["Age"] > 100), "Age"] = pd.NA

# then validate if the minimum and maximum age values are much more specific
display(customer_cleaned["Age"].describe())
print("\nMissing Ages:", customer_cleaned["Age"].isnull().sum())
```

```python
# fill missing age values with median to preserve customer records while minimixing the influence of extreme values
customer_cleaned["Age"] = customer_cleaned["Age"].fillna(customer_cleaned["Age"].median())

# verify
print("Missing Values:", customer_cleaned["Age"].isnull().sum())

print("Numerical Statistics of Age Column after Filling Missing and Invalid Values:")
display(customer_cleaned["Age"].describe())
```

##### income level column

```python
# inspect missing values
print(customer_cleaned["IncomeLevel"].value_counts(dropna=False))
```

285 missing values from incomelevel column cannot be replaced by 'high', despite being the mode, since it would artificially increase the high income group which could affect later eda and visualizations

```python
# fill missing values with 'unkown'
customer_cleaned["IncomeLevel"] = customer_cleaned["IncomeLevel"].fillna("Unkown")

# verify
print(customer_cleaned["IncomeLevel"].value_counts(dropna=False))
```
