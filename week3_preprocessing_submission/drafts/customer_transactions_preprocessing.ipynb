```python
# Import libraries for data preprocessing and analysis
import pandas as pd
import numpy as np
```

```python
# Load the contaminated customer transactions dataset
transactions = pd.read_csv("customer_transactions_contaminated.csv")
```

```python
# Inspect the first five rows of the dataset
transactions.head()
```

```python
# Check the number of rows and columns in the dataset
transactions.shape
```

```python
# Review the data types of all columns
transactions.dtypes
```

```python
# Identify the number of missing values in each column
transactions.isnull().sum()
```

```python
# Check the number of duplicate rows in the dataset
transactions.duplicated().sum()
```

```python
# Remove duplicate rows and create a cleaned copy of the dataset
transactions_cleaned = transactions.drop_duplicates().copy()
```

```python
# Check the dimensions after removing duplicate rows
transactions_cleaned.shape
```

```python
# Inspect the most frequent values in the Amount column,
# including missing values
transactions_cleaned["Amount"].value_counts(dropna=False).head(20)
```

```python
# Convert the Amount column to numeric values
# Invalid text values such as "Free" are converted to missing values
transactions_cleaned["Amount"] = pd.to_numeric(
    transactions_cleaned["Amount"],
    errors="coerce"
)
```

```python
# Verify that the Amount column has been converted to a numeric data type
transactions_cleaned["Amount"].dtype
```

```python
# Identify transaction records with negative Amount values
transactions_cleaned[transactions_cleaned["Amount"] < 0]
```

```python
# Replace invalid negative transaction amounts with missing values
transactions_cleaned.loc[
    transactions_cleaned["Amount"] < 0,
    "Amount"
] = np.nan
```

```python
# Count the remaining missing values in the Amount column
transactions_cleaned["Amount"].isnull().sum()
```

```python
# Calculate the median of the valid transaction amounts
transactions_cleaned["Amount"].median()
```

```python
# Replace missing and invalid Amount values with the median value
transactions_cleaned["Amount"] = transactions_cleaned["Amount"].fillna(504.14)
```

```python
# Verify that no missing values remain in the Amount column
transactions_cleaned["Amount"].isnull().sum()
```

```python
# Inspect product categories and identify missing category values
transactions_cleaned["ProductCategory"].value_counts(dropna=False)
```

```python
# Replace missing product categories with "Unknown"
transactions_cleaned["ProductCategory"] = transactions_cleaned["ProductCategory"].fillna("Unknown")
```

```python
# Verify that no missing values remain in ProductCategory
transactions_cleaned["ProductCategory"].isnull().sum()
```

```python
# Inspect the most common transaction dates
transactions_cleaned["TransactionDate"].value_counts().head(10)
```

```python
# Identify transaction dates that use the DD/MM/YYYY format
transactions_cleaned[
    transactions_cleaned["TransactionDate"].str.contains("/", na=False)
]["TransactionDate"].value_counts().head(20)
```

```python
# Convert mixed date formats into a consistent datetime format
transactions_cleaned["TransactionDate"] = pd.to_datetime(
    transactions_cleaned["TransactionDate"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)
```

```python
# Verify that TransactionDate is stored as a datetime data type
transactions_cleaned["TransactionDate"].dtype
```

```python
# Check whether any transaction dates became missing after conversion
transactions_cleaned["TransactionDate"].isnull().sum()
```

```python
# Perform a final quality check on the cleaned dataset

print("FINAL DATASET CHECK")
print("-------------------")

print("\nDataset Shape:")
print(transactions_cleaned.shape)

print("\nData Types:")
print(transactions_cleaned.dtypes)

print("\nMissing Values:")
print(transactions_cleaned.isnull().sum())

print("\nDuplicate Rows:")
print(transactions_cleaned.duplicated().sum())
```

```python
# Save the finalized preprocessed dataset as a new CSV file
# The original contaminated dataset remains unchanged
transactions_cleaned.to_csv(
    "customer_transactions_preprocessed.csv",
    index=False
)

print("Preprocessed dataset saved successfully!")
```
