```python
import pandas as pd
```

```python
# load dataset

transactions = pd.read_csv("customer_transactions_contaminated.csv")
```

## Inspect and Clean Customer Transactions Dataset

```python
# inspect dataset

print("First 20 rows: ")
display(transactions.head(20))

print("\nDataset Information: ")
display(transactions.info())

print("\nDataset Dimensions: ")
display(transactions.shape)

print("\nDataset Columns: ")
display(transactions.columns)

print("\nDataset Numerical Statistics: ")
display(transactions.describe())
```

##### handling exact duplicate values

```python
# check duplicates
duplicate_transactions = transactions.duplicated().sum()

print("Number of Exact Duplicate Rows: ")
print(duplicate_transactions)

# inspect duplicates
display(transactions[transactions.duplicated(keep=False)])
```

```python
# remove duplicates
transaction_cleaned = transactions.drop_duplicates().copy()

# verify
print("Duplicates Remaining:")
print(transaction_cleaned.duplicated().sum())
```

#### handling missing and invalid values

##### amount column

```python
# check missing values
missing_values = transaction_cleaned.isnull().sum()
print("Missing Values:")
print(missing_values)
```

```python
# inspect data types and check numerical statistics
print("Data Types:")
print(transaction_cleaned.dtypes)

print("\nNumerical Statistics:")
display(transaction_cleaned.describe())
```

```python
# check categorical inconsistencies
for column in transaction_cleaned.select_dtypes(include="object").columns:
    print(f"\n{column}")
    print(transaction_cleaned[column].value_counts(dropna=False))
```

##### amount column

```python
# convert amount from string to numeric

transaction_cleaned["Amount"] = pd.to_numeric(
    transaction_cleaned["Amount"],
    errors="coerce"
)

# check numerical statistics

print("Numerical Statistics of Amount Column:")
display(transaction_cleaned["Amount"].describe())

# check again number of missing values

print("Number of Missing Values:")
print(transaction_cleaned["Amount"].isnull().sum())
```

### invalid amount values are dealt with prior to filling or dropping missing values

```python
# count how many invalid amount values are in the dataset

invalid_amount = transaction_cleaned[
    (transaction_cleaned["Amount"] < 0) |
    (transaction_cleaned["Amount"] == 0)
]

print("Invalid Amount Values:")
print(len(invalid_amount))

display(invalid_amount)
```

```python
# convert invalid amount values to missing values

transaction_cleaned.loc[
    (transaction_cleaned["Amount"] < 0) |
    (transaction_cleaned["Amount"] == 0),
    "Amount"
] = pd.NA

# validate

print("Missing Amount Values:")
print(transaction_cleaned["Amount"].isnull().sum())

print("\nAmount Statistics After Removing Invalid Values:")
display(transaction_cleaned["Amount"].describe())
```

```python
# fill missing amount values with the median

transaction_cleaned["Amount"] = transaction_cleaned["Amount"].fillna(
    transaction_cleaned["Amount"].median()
)

# verify

print("Missing Amount Values After Filling:")
print(transaction_cleaned["Amount"].isnull().sum())

print("\nAmount Statistics After Cleaning:")
display(transaction_cleaned["Amount"].describe())
```

##### product category column

```python
# inspect missing values

print("Product Category Values:")
display(transaction_cleaned["ProductCategory"].value_counts(dropna=False))
```

#### missing product category values are replaced with 'Unknown' because the original category is unavailable

```python
# fill missing values with 'Unknown'

transaction_cleaned["ProductCategory"] = transaction_cleaned[
    "ProductCategory"
].fillna("Unknown")

# verify

print("Product Category Values After Cleaning:")
display(transaction_cleaned["ProductCategory"].value_counts(dropna=False))
```

##### transaction date column

```python
# convert transaction date to datetime

transaction_cleaned["TransactionDate"] = pd.to_datetime(
    transaction_cleaned["TransactionDate"],
    errors="coerce",
    format="mixed",
    dayfirst=True
)

# verify

print("Transaction Date Data Type:")
print(transaction_cleaned["TransactionDate"].dtype)
```

```python
# check missing or invalid transaction dates

print("Missing Transaction Dates:")
print(transaction_cleaned["TransactionDate"].isnull().sum())

print("\nTransaction Date Range:")
print("Minimum Date:", transaction_cleaned["TransactionDate"].min())
print("Maximum Date:", transaction_cleaned["TransactionDate"].max())
```

##### payment method column

```python
# inspect payment methods

print("Payment Method Values:")
display(transaction_cleaned["PaymentMethod"].value_counts(dropna=False))
```

##### customer id column

```python
# investigate duplicate customer id values

duplicate_customer_ids = transaction_cleaned[
    transaction_cleaned.duplicated(
        subset="CustomerID",
        keep=False
    )
]

print("Duplicate Customer ID Records:")
print(len(duplicate_customer_ids))

display(duplicate_customer_ids.head(20))
```

##### transaction id column

```python
# investigate duplicate transaction id values

duplicate_transaction_ids = transaction_cleaned[
    transaction_cleaned.duplicated(
        subset="TransactionID",
        keep=False
    )
]

print("Duplicate Transaction ID Records:")
print(len(duplicate_transaction_ids))

display(
    duplicate_transaction_ids.sort_values("TransactionID")
)
```

#### final dataset verification

```python
# verify cleaned dataset

print("Cleaned Dataset Information:")
transaction_cleaned.info()

print("\nCleaned Dataset Dimensions:")
print(transaction_cleaned.shape)

print("\nRemaining Missing Values:")
display(transaction_cleaned.isnull().sum())

print("\nRemaining Exact Duplicates:")
print(transaction_cleaned.duplicated().sum())
```

### Important Note

> **The customer transactions data underwent preprocessing to improve its quality and consistency. Exact duplicate records were removed, invalid and missing Amount values were addressed, missing ProductCategory entries were assigned as "Unknown," and TransactionDate was converted into a standardized date format. After cleaning, the dataset contained 3,015 records across 6 columns, with no remaining missing values or exact duplicate rows.**
>
> **However, duplicate CustomerID values were retained because a single customer can have multiple transactions. Duplicate TransactionID values were examined but were not removed automatically, since further verification is necessary to determine whether they represent genuine records or duplicate transactions.**

```python
# save cleaned dataset

transaction_cleaned.to_csv(
    "customer_transactions_preprocessed.csv",
    index=False
)
```
